```python
import pandas as pd

usaid = pd.read_csv("../data/raw/usaid/Raw_Data.csv")
usaid = usaid.rename(columns={usaid.columns[0]: "ID"})

print("PAYS DE LIVRAISON :")
print(dict(usaid["Country"].value_counts().head(8)))

print("\nTRANSPORT :")
print(dict(usaid["Shipment Mode"].value_counts(dropna=False)))

print("\nUSINES (top 8) :")
print(dict(usaid["Manufacturing Site"].value_counts().head(8)))

prevu = pd.to_datetime(usaid["Scheduled Delivery Date"], format="%d-%b-%y", errors="coerce")
reel  = pd.to_datetime(usaid["Delivered to Client Date"], format="%d-%b-%y", errors="coerce")
usaid["retard_jours"] = (reel - prevu).dt.days

print("\nPÉRIODE :", reel.min().date(), "→", reel.max().date())
print("Livraisons en retard :", round((usaid["retard_jours"] > 0).mean() * 100, 1), "%")
print("Retard moyen quand il y a retard :", round(usaid.loc[usaid["retard_jours"] > 0, "retard_jours"].mean(), 1), "jours")
```

```python
import pandas as pd
from ftfy import fix_text

usaid = pd.read_csv("../data/raw/usaid/Raw_Data.csv")
usaid = usaid.rename(columns={usaid.columns[0]: "ID"})

colonnes_texte = usaid.select_dtypes(include="object").columns
for col in colonnes_texte:
    usaid[col] = usaid[col].map(lambda x: fix_text(x) if isinstance(x, str) else x)

print("Pays :", [p for p in usaid["Country"].unique() if "Ivoire" in str(p)])

usines = usaid["Manufacturing Site"].value_counts()
print("Nombre d'usines différentes :", len(usines))
usines.to_csv("../data/raw/usaid/usines.csv")
```
